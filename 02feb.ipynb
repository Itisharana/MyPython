## Pattern


```python
for i in range(1,10):
    print('*')
```

```python
for i in range(1,10):
    for j in range(1, 10):
        print('*', end = ' ' )
```

```python
for i in range(1,10):
    for j in range(1, 10):
        print('*', end = ' ' )
    print()
```

```python
for i in range(1,10):
    for j in range(i):
        print('*', end = ' ' )
    print()
```

```python
for i in range(1,10):
    for j in range(j):
        print('*', end = ' ' )
    print()
```

```python
for i in range(10,0, -1):
    for j in range(i):
        print('*', end = ' ' )
    print()
```

```python
for i in range(1,10):
    for j in range(i):
        print('*', end = ' ' )
    print()

for i in range(10,0, -1):
    for j in range(i):
        print('*', end = ' ' )
    print()
```

```python
n = 5
for i in range(1 , n +1):
    for space in range(n-i):
        print(' ', end = ' ')
    for star in range(i):
        print('*', end = ' ')
    print()
```

```python
n = 5
for i in range(n, 0, -1):
    for space in range(n-i):
        print(' ', end = ' ')
    for star in range(i):
        print('*', end = ' ')
    print()
```

```python
n = 6
m=5
for i in range(1 , n +1):
    for space in range(n-i):
        print(' ', end = ' ')
    for star in range(i):
        print('*', end = ' ')
    print()
for i in range(m, 0, -1):
    for space in range(n-i):
        print(' ', end = ' ')
    for star in range(i):
        print('*', end = ' ')
    print()
```

```python
for i in range(1,5):
    for j in range(1, i+1):
        print(i, end = ' ')
    print()
```

```python
num = 1
for i in range(1,5):
    for j in range(i):
        print(num, end = ' ')
        num +=1
    print()
```
