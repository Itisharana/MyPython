## oops abstract method
Blueprint or structure. fucntions defined are compulsary to be in the class otherwise error

```python
from abc import ABC, abstractmethod
```

```python
class employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass 
    @abstractmethod
    def details(self):
        pass
```

```python
class employee(employee):
    def __init__(self):
        self.name = input('enter name')
        self.salary = int(input('enter salary'))
        self.pf=self.salary*0.20
        self.calculate_salary()
    def calculate_salary(self):
        self.net_salary = self.salary+self.pf
        print(self.net_salary)
    def details(self):
        print(f"name of employee is {self.name} and its salary {self.net_salary}")
```

```python
user1 = employee()
```

```python
user1.calculate_salary()
user1.details()
```

```python
user1.details()
```

## Try except

```python
a = int(input('num1'))
b = int(input('num2'))
print(a/b)
```

```python
try:
    a = int(input('num1'))
    b = int(input('num2'))
    print(a/b)
except(ZeroDivisionError):
    print('number cannot be divided by 0')
except(ValueError):
    print('enter number only')
finally:
    print('program end successfully')
```
