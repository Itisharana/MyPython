```python
age = int(input())
if age>100 and age<0:
    if age >= 18 and age <= 30:
        print('adult')
    elif age >= 31 and age <= 50:
        print('mature')
    elif age >= 51 and age <= 60:
        print('old man')
    elif age >= 61:
        print('retired')
    else:
        print('not adult')
else:
    print('enter valid age')
```

```python
number = int(input())
if number == 0:
    print('neutral')
elif number%2 == 0 and number > 0:
    print('even')
else:
    print('odd')
```

```python
choice = int(input())
match choice:
    case 1:
        print('one')
    case 2:
        print('two')
    case 3:
        print('three')
    case _:
        print('invalid')
```

```python
aa = int(input('num1'))
bb = int(input('num2'))
op = input('operation')
match op:
    case '+':
        print(aa+bb)
    case '-':
        print(aa-bb)
    case '*':
        print(aa*bb)
    case '/':
        print(aa/bb)
    case _:
        print('invalid')
```

```python
day = input('day')
match day:
    case 'sat'| 'sun':
        print('weekend')    
    case 'mon'| 'tue'|'wed'|'thrs'|'fri':
        print('weekday')
    case _:
        print('invalid')
```
